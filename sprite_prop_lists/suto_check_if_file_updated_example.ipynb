```python
import os
import time
import threading
import ipywidgets as widgets
from IPython.display import display

# 1. Create the output widget container
output_widget = widgets.Output()

FILE_PATH = "data.txt"

def watch_file(out_widget, path):
    last_mtime = 0
    # Keep running as long as the Voilà session is alive
    while True:
        if os.path.exists(path):
            current_mtime = os.path.getmtime(path)
            
            # If the file's modification time changes, refresh the output
            if current_mtime > last_mtime:
                last_mtime = current_mtime
                
                with out_widget:
                    out_widget.clear_output(wait=True)
                    with open(path, "r") as f:
                        print(f"File updated at {time.ctime(current_mtime)}:")
                        print(f.read())
                        
        time.sleep(1) # Poll every 1 second

# 2. Render the widget to the dashboard layout
display(output_widget)

# 3. Start the background thread
thread = threading.Thread(target=watch_file, args=(output_widget, FILE_PATH), daemon=True)
thread.start()
```
