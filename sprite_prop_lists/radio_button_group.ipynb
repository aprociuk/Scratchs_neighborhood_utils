```python
import ipywidgets as widgets
from IPython.display import display

# Define questions and their options
questions = {
    "Question 1": ["Option A", "Option B", "Option C"],
    "Question 2": ["Yes", "No"],
    "Question 3": ["Low", "Medium", "High"],
}

# Create a list of RadioButtons widgets
radio_groups = {}
for q, opts in questions.items():
    radio_groups[q] = widgets.RadioButtons(
        options=opts,
        description=q,
        disabled=False,
    )

# Stack them vertically
ui = widgets.VBox(list(radio_groups.values()))
display(ui)
```
