```python
import pandas as pd
from ipydatagrid import DataGrid

# 1. Create a sample dataset
data = {
    "Product": ["Laptop", "Smartphone", "Tablet", "Headphones"],
    "Price": [999.99, 599.50, 349.99, 149.00],
    "In Stock": [True, True, False, True],
    "Rating": [4.7, 4.5, 4.1, 4.6]
}
df = pd.DataFrame(data)

# 2. Pass the DataFrame to the DataGrid constructor
grid = DataGrid(
    df, 
    selection_mode="cell",  # Allows cell-by-cell selection grid
    editable=True           # Enables direct editing in the UI
)

# 3. Print/Display the grid in the notebook
grid
```
