```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Create a global mutable object
shared_state = {
    "counter": 0,
    "history": []
}

# Create widgets
btn_increment = widgets.Button(description="Increment Counter", button_style="success")
btn_status = widgets.Button(description="Check Status", button_style="info")
output = widgets.Output()

# 2. Callback function that mutates the global object
def on_increment_clicked(b):
    shared_state["counter"] += 1
    shared_state["history"].append(f"Incremented to {shared_state['counter']}")
    
    with output:
        clear_output(wait=True)
        print(f"Action triggered! Counter is now: {shared_state['counter']}")

# 3. Another callback function that reads/mutates the same global object
def on_status_clicked(b):
    with output:
        clear_output(wait=True)
        print("--- Current Global State ---")
        print(f"Total Counts: {shared_state['counter']}")
        print(f"Log History: {shared_state['history']}")

# Bind the callbacks to the buttons
btn_increment.on_click(on_increment_clicked)
btn_status.on_click(on_status_clicked)

# Display the user interface
display(widgets.HBox([btn_increment, btn_status]), output)
```
