```python
import asyncio
import ipywidgets as widgets
from IPython.display import display

# Google Gemini response to question:
# python can you programmatically issue a command to open an app and capture a signal when the app is closed within a voila dashboard
# Use sys.platform to dynamically change the executable app based on OS/Platform

# 1. Create UI Elements
button = widgets.Button(
    description="Launch App", 
    button_style="primary", 
    icon="rocket"
)
status_label = widgets.Label(value="Status: Idle")

# 2. Define the Async Process Launcher
async def launch_and_wait(button_obj):
    button.disabled = True
    status_label.value = "Status: App is running..."
    
    try:
        # Replace 'notepad.exe' (Windows) with your app, e.g., 'vlc' or 'gedit' (Linux/Mac)
        # Using asyncio.create_subprocess_exec prevents blocking the Voila server
        process = await asyncio.create_subprocess_exec(
            'notepad.exe', 
            stdout=asyncio.subprocess.DEVNULL,
            stderr=asyncio.subprocess.DEVNULL
        )
        
        # Wait for the specific process exit signal
        return_code = await process.wait()
        
        status_label.value = f"Status: App closed successfully (Exit Code: {return_code})"
    except Exception as e:
        status_label.value = f"Status: Failed to launch app ({str(e)})"
    finally:
        button.disabled = False

# 3. Bind Button Click to the Event Loop
def on_button_click(b):
    # Schedules the async task in Voila's existing event loop
    asyncio.create_task(launch_and_wait(b))

button.on_click(on_button_click)

# 4. Render Layout
display(widgets.VBox([button, status_label]))
```
