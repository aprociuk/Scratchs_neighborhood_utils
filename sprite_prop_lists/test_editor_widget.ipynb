```python
import ipywidgets as widgets
from IPython.display import display, HTML

# 1. Force monospace font for code/text editing look
display(HTML("<style> textarea { font-family: monospace; } </style>"))

# 2. Load initial file content
file_path = "sample.txt"
try:
    with open(file_path, "r", encoding="utf-8") as f:
        initial_text = f.read()
except FileNotFoundError:
    initial_text = "# Type your text here..."

# 3. Create Textarea widget
text_editor = widgets.Textarea(
    value=initial_text,
    placeholder="Edit file contents here...",
    description="File:",
    disabled=False,
    layout=widgets.Layout(width="100%", height="300px"),
)

# 4. Create Save Button
save_button = widgets.Button(
    description="Save File",
    button_style="success",
    icon="save",
)

status_output = widgets.Output()


def save_file_callback(b):
    with status_output:
        status_output.clear_output()
        try:
            with open(file_path, "w", encoding="utf-8") as f:
                f.write(text_editor.value)
            print(f"Successfully saved to {file_path}")
        except Exception as e:
            print(f"Error saving file: {e}")


save_button.on_click(save_file_callback)

# Display in dashboard
display(text_editor, save_button, status_output)
```
